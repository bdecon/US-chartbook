### OECD Indicators

```python
import sys
sys.path.append('../src')

import uschartbook.config

from uschartbook.config import *
from uschartbook.utils import *
import io
```

```python
rn = {'DNK': 'Denmark', 'FIN': 'Finland', 'NOR': 'Norway', 'SWE': 'Sweden'}
web_dir = '/home/brian/Documents/bdecon.github.io/files'
```

### Productivity (GDP per hour worked)

```python
url = ('https://sdmx.oecd.org/public/rest/data/'
       'OECD.SDD.TPS,DSD_PDB@DF_PDB_LV,1.0/'
       'DNK+FIN+NOR+SWE+USA.A.GDPHRS..USD_PPP_H.Q...?'
       'startPeriod=2020&format=csvfilewithlabels')

r = requests.get(url)
res = pd.read_csv(io.StringIO(r.text))
data = res.set_index(['REF_AREA', 'TIME_PERIOD'])['OBS_VALUE'].unstack().T
data.index = pd.to_datetime(data.index, format='%Y')
data.index.name = ''
data.columns.name = ''

result = data.dropna().iloc[-1].round(1).rename(rn)
result.name = 'Value'
result.to_csv(f'{web_dir}/lprodbar.csv', index_label='Country')
print(f'Productivity ({data.dropna().index[-1].year}):')
result
```
